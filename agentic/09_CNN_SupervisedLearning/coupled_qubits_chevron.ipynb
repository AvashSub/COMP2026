# Coupled qubits: reading the coupling from chevron patterns

**The system.** Two superconducting qubits (transmons) with a direct exchange coupling $g$. Each transmon is a slightly anharmonic oscillator: its levels $|0\rangle, |1\rangle, |2\rangle$ are not evenly spaced, and $|2\rangle$ sits $\alpha$ off from where an evenly spaced ladder would put it ($\alpha < 0$, the *anharmonicity*). Keeping three levels per qubit gives
$$H = \sum_{i=1,2}\Big[\omega_i\, n_i + \tfrac{\alpha_i}{2}\, n_i(n_i - 1)\Big] + g\,\big(a_1^\dagger a_2 + a_1 a_2^\dagger\big),$$
where $a_i$ lowers qubit $i$ by one level and $n_i = a_i^\dagger a_i$ counts its excitations. Qubit 2 sits at a fixed frequency $\omega_2$. Qubit 1's frequency $\omega_1$ is tunable (on hardware, with a flux pulse).

**Excitation number is conserved.** The coupling only moves one excitation from one qubit to the other, so $H$ splits into independent blocks:
- **1 excitation**, $\{|10\rangle, |01\rangle\}$, coupled by $g$. This is the *swap* process: the excitation hops back and forth between the qubits.
- **2 excitations**, $\{|11\rangle, |20\rangle, |02\rangle\}$, coupled by $\sqrt{2}\,g$. The $\sqrt2$ comes from $a^\dagger|1\rangle = \sqrt2\,|2\rangle$.

**Two chevrons.** Prepare a state, set $\omega_1$, let the qubits interact for a time $t$, then measure. Sweeping $(\omega_1, t)$ gives two images:

| Chevron | Start | Measure | Centered at | Fringe frequency on resonance |
|---|---|---|---|---|
| Swap (iSWAP) | $\vert 10\rangle$ | $P(\vert 01\rangle)$ | $\omega_1 = \omega_2$ | $2g$ |
| CZ | $\vert 11\rangle$ | $P(\vert 02\rangle)$ | $\omega_1 = \omega_2 + \alpha$ | $2\sqrt2\, g$ |

The swap chevron has exactly the single-qubit Rabi form, with $\Omega \to 2g$ and $\Delta \to \delta = \omega_1 - \omega_2$:
$$P(|01\rangle) = \frac{4g^2}{4g^2 + \delta^2}\,\sin^2\!\Big(\tfrac{\sqrt{4g^2+\delta^2}\;t}{2}\Big).$$
The CZ chevron needs the 3-level block, and is computed numerically.

**What we extract.**
- $g$, the coupling: the fringe spacing. It sets the gate times, $t = \pi/(2g)$ for a full iSWAP and $\pi/(\sqrt2 g)$ for a CZ.
- $\omega_2$: the center of the swap chevron.
- $\alpha$: how far the CZ chevron is shifted from the swap chevron.
- the decay time: how fast the fringes fade, as in the single-qubit case.

The CNN gets both chevrons as a **2-channel image** and regresses all of these at once, with the same noise model as before. Consistency checks:
- the 1-excitation block must reproduce the analytic swap formula;
- the CZ fringes must oscillate exactly $\sqrt2$ times faster than the swap fringes.

*Conventions:* $\hbar = 1$; all frequencies are angular and in the same units, with $t$ in the inverse of those units.
